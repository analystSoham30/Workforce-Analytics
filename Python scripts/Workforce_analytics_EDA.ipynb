```python
def identify_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3-Q1

    lower_bound = Q1-1.5*IQR
    upper_bound = Q3+1.5*IQR

    return df[(df[column] < lower_bound) | (df[column] > upper_bound)]
```

```python
imp_cols = employee_df.select_dtypes(include = [np.number]).columns
imp_cols = imp_cols.drop(['Employee_ID','Remote_Work_Frequency'])
imp_cols
```

```python
for col in imp_cols:
    outliers = identify_outliers(employee_df,col)
    print(outliers)
```
